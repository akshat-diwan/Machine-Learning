# Spam Mail Prediction

**Import Dependancy**

```python
# Import Dependancy
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split

from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report, precision_score, recall_score
from sklearn.feature_extraction.text import TfidfVectorizer
```

**Import the mail data**

```python
df = pd.read_csv('/content/drive/MyDrive/Colab Notebooks/Toppers Acadmy Data/data/Machine Learning/Chapter 5:- Natural language processing(NLP)/Project/Copy of mail_data.csv')
```

```python
df.head()
```

```python
df['Category'].value_counts()
```

### Dealing with null values

**Replacing null values with null string**

```python
df.isnull().sum()
```

### Lable Encoding

*   ham	converted as 1
*   spam converted as 0



```python
df['label'] = df['Category'].apply(lambda x: 1 if x == 'spam' else 0)
df['label'].value_counts()
```

### Train Test Split

**Sepret data as x and y**

```python
x = df['Message']
y = df['label']
```

```python
x
```

```python
y
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=3)
```

### Feature Extraction

**convert text data into numerical data**

```python
feature_extraction = TfidfVectorizer(min_df=1, stop_words='english', lowercase=True)
x_train = feature_extraction.fit_transform(x_train)
x_test = feature_extraction.transform(x_test)

```

### Create model

**Logistic regression**

```python
model = LogisticRegression()
```

```python
model.fit(x_train,y_train)
```

```python
train_pred = model.predict(x_train)
test_pred = model.predict(x_test)
```

```python
train_pred[:5]
```

```python
y_train.head()
```

```python
test_pred[:5]
```

```python
y_test.head()
```

### Eveluating the training model

**Prediction score on training data**

```python
print('train accu :- ',accuracy_score(y_train,train_pred))
print('train confusion matrix :- \n',confusion_matrix(y_train,train_pred))
print('train precision :- ',precision_score(y_train,train_pred))
print('train recall :- ',recall_score(y_train,train_pred))
```

```python
print('Test acc :- ',accuracy_score(y_test,test_pred))
print('Test confusion matrix :- \n',confusion_matrix(y_test,test_pred))
print('Test precision :- ',precision_score(y_test,test_pred))
print('Test recall :- ',recall_score(y_test,test_pred))
```

### Building a Predictive System

```python
input_data = ["i am going to delhi."]
data = feature_extraction.transform(input_data)
prediction = model.predict(data)
if prediction[0] == 1:
  print('Spam mail')
else:
  print('ham mail')
```
